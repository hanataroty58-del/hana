```python
import pandas as pd
import sqlite3
conn = sqlite3.connect('/content/download.db')
```

```python

def run(query) :
    return pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    members.*,
    checkouts.checkout_id,
    checkouts.book_id,
    checkouts.checkout_date
FROM members
JOIN checkouts ON members.member_id = checkouts.member_id;
"""
run(query)
```

```python
query='''SELECT *
FROM books
WHERE author LIKE '%mo%';'''
run(query)
```

```python
query='''SELECT
    members.member_id,
    members.first_name,
    members.last_name,
    COUNT(checkouts.checkout_id) AS total_checkouts
FROM members
JOIN checkouts ON members.member_id = checkouts.member_id
GROUP BY
    members.member_id,
    members.first_name,
    members.last_name'''
run(query)
```

```python
query='''SELECT
    books.title,
    COUNT(checkouts.checkout_id) AS times_checked_out
    FROM
    books
    JOIN
    checkouts ON books.book_id = checkouts.book_id
    GROUP BY
    books.book_id,
    books.title ORDER BY
    times_checked_out DESC
    LIMIT 5;'''
run(query)
```

```python
query='''SELECT
    members.member_id,
    members.first_name,
    members.last_name,
    COUNT(checkouts.checkout_id) AS books_borrowed
FROM
    members
JOIN
    checkouts ON members.member_id = checkouts.member_id
GROUP BY
    members.member_id,
    members.first_name,
    members.last_name
ORDER BY
    books_borrowed DESC
LIMIT 10;'''
run(query)
```

```python
query='''SELECT
    members.neighborhood,
    members.member_id,
    checkouts.checkout_id,
    checkouts.checkout_date
FROM
    members
JOIN
    checkouts ON members.member_id = checkouts.member_id
WHERE
    members.neighborhood = 'Zamalek'
ORDER BY
    checkouts.checkout_date DESC
LIMIT 10 OFFSET 10;'''
```

```python
run(query)
```
