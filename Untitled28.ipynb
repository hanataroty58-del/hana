```python
import json
import sqlite3
import pandas as pd









```

```python
df_html_checkouts = pd.read_html("download.html")[0]
df_html_checkouts.rename(
    columns={
        "Member ID": "member_id",
        "Book ID": "book_id",
        "Checkout Date": "checkout_date",
    },
    inplace=True,
)
```

```python
df_json_books = pd.read_json("download.json")
```

```python
conn = sqlite3.connect("download.db")
df_db_checkouts = pd.read_sql_query("SELECT * FROM checkouts", conn)
df_db_books = pd.read_sql_query("SELECT * FROM books", conn)
df_db_members = pd.read_sql_query("SELECT * FROM members", conn)
```

```python
df_complete_books = pd.merge(df_db_books, df_json_books, on="book_id")
df_complete_checkouts = pd.concat(
    [df_db_checkouts, df_html_checkouts]
)
```

```python
df_master_dataset = (
    df_complete_checkouts.merge(df_db_members, on="member_id")
    .merge(df_complete_books, on="book_id")
)
```

```python

df_master_dataset.to_csv("task1_combined_data.csv", index=False)
```
