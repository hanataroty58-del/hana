```python
import pandas as pd
import sqlite3
conn = sqlite3.connect('/content/download.db')

```

```python
def run(query) :
    return pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    members.*,
    checkouts.checkout_id,
    checkouts.book_id,
    checkouts.checkout_date
FROM members
JOIN checkouts ON members.member_id = checkouts.member_id;
"""
run(query)
```

```python
df_db = pd.read_sql_query(query, conn)
```

```python
df_db
```

```python
df_json=pd.read_json("/content/download.json")
```

```python
df_json
```

```python
df_html_list=pd.read_html("/content/download.html")
df_html = df_html_list[0]
```

```python
df_html
```

```python
df_merged = pd.merge(df_db, df_json, on="book_id")
```

```python
df_html.columns = df_html.columns.str.lower().str.replace(' ', '_')
```

```python
df_html
```

```python
df_final = pd.merge(df_merged, df_html, on="member_id")
```

```python
df_final.to_csv("task1_combined_data.csv", index=False)
```
